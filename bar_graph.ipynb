```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Data based on the XGBoost Feature Importance from the manuscript
# The features are ordered from most important to least important (top to bottom on the y-axis)
# Mapping: f2 -> decgeo, f0 -> rageo, f1 -> vgeo, f3 -> sol_lon, f5 -> elev, f4 -> azim

# Features (in descending order of importance)
feature_labels = [
    r'$\text{Geocentric Declination}$',  # Geocentric Declination (f2)
    r'$\text{Geocentric Right Ascension}$',  # Geocentric Right Ascension (f0)
    r'$\text{Geocentric Velocity}$',   # Geocentric Velocity (f1)
    r'$\text{Solar Longitude}$', # Solar Longitude (f3)
    r'$\text{Local Elevation}$',   # Local Elevation (f5)
    r'$\text{Local Azimuth}$'    # Local Azimuth (f4)
]

# Importance scores (corresponding to the ordered features)
importance_scores = [
    4595.0,
    4260.0,
    4126.0,
    2999.0,
    247.0,
    230.0
]

# The plot should display the features in the order of the original figure (f2 at top, f4 at bottom).
# We reverse the order for plotting because matplotlib's barh plots from bottom-up.
importance_scores.reverse()
feature_labels.reverse()

# 2. Plotting the Bar Chart
sns.set_style("whitegrid")
plt.figure(figsize=(10, 6))

# Create the horizontal bar plot
bars = plt.barh(feature_labels, importance_scores, color='#1f77b4')

# 3. Customizing the Plot
plt.xlabel('Importance score', fontsize=12)
plt.ylabel('Features', fontsize=12)
# plt.title('Figure 5: Feature Importance (Revised with Scientific Labels)', fontsize=14, pad=20)

# Add the score labels at the end of each bar
for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 100,  # x position, slightly offset from the bar end
        bar.get_y() + bar.get_height() / 2, # y position (center of the bar)
        f'{width:.1f}',  # Score text, formatted to one decimal place
        ha='left',
        va='center',
        fontsize=10
    )

# Adjust x-axis limits and ticks
plt.xlim(0, max(importance_scores) * 1.15)
plt.xticks(range(0, 5001, 1000))

# Remove the top and right spines for a cleaner look
sns.despine(trim=True)

# Save the figure (optional)
# plt.savefig('revised_feature_importance_plot.png', bbox_inches='tight', dpi=300)

plt.tight_layout()
plt.show()
```
